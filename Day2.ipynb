```python
class Student:
    def __init__(self,usn,name,age):
        #instance variable
       self.usn = usn
       self.name = name
       self.age = age
   # instance method access instance variable
    def show (self):
       print('USN',self.usn,'Name',self.name,'Age',self.age)
       # instance method to modify instance variablr
    def update(self,usn,age):
          self.usn = usn
          self.age = age
    
```

```python
# create an object
stu =Student(40,'rms',21)
stu.show()
```

```python
stu.update(12,38)
stu.show()
```
