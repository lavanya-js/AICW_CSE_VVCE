```python
from datetime import date
```

```python

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def calculate_age(cls, name, birthyear):
        return cls(name, date.today().year - birthyear)

    def show(self):
        print(self.name + "'s age is " + str(self.age))


age = Student.calculate_age("Lavanya", 2006)
age.show()
```

```python
class Person:
    def Person_info(self, name, age):
        print('Inside person class - parent class')
        print('Name:', name, 'Age:', age)


class Company:
    def Company_info(self, company_name, location):
        print('Inside company class - parent class')
        print('Company:', company_name, 'Location:', location)


class Employee(Person, Company):
    def Employee_info(self, salary, skill):
        print('Inside employee class - child class')
        print('Salary:', salary, 'Skills:', skill)


# Creating object
e = Employee()

# Calling methods
e.Person_info('Rahul', 22)
e.Company_info('Google', 'Bangalore')
e.Employee_info(50000, 'Python')
```

```python
class Car:
    def __init__(self,brand,model):
        self.brand = brand
        self.model = model
    def move(self):
        print("Drive!")
class Boat:
    def __init__(self,brand,model):
        self.brand = brand
        self.model = model
    def move(self):
        print("Sail!")
class Plane:
    def __init__(self,brand,model):
        self.brand = brand
        self.model = model
    def move(self):
        print("Fly!!")
```

```python
#polymorphism with inheritance
class Bird:
    def intro(self):
        print('There are many type of birds!')
    def flight(self):
        print('Most of the birds can fly but some cannot!')
class sparrow(Bird):
    def flight(self):
        print("Sparrow can fly!")
class ostrich(Bird):
    def flight(self):
        print("Ostrich cannot fly!")        
```

```python
ob_bird = Bird()
ob_bird.intro()
ob_bird.flight()
```

```python
ob_sparrow = sparrow()
ob_sparrow.flight
```
